```python
goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

visited = set()

def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i+1], state[i+2])
    print()

def dfs(state, path):

    if state == goal:
        print("Goal State Reached!")
        print_puzzle(state)
        print("Moves:", path)
        return True

    visited.add(state)

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [
        ("Down", 1, 0),
        ("Up", -1, 0),
        ("Right", 0, 1),
        ("Left", 0, -1)
    ]

    for move, dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = list(state)

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            new_state = tuple(new_state)

            if new_state not in visited:

                print("Move:", move)
                print_puzzle(new_state)

                if dfs(new_state, path + [move]):
                    return True

    return False

print("Initial State:")
print_puzzle(start)

if not dfs(start, []):
    print("Solution not found")
```

```python
goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i+1], state[i+2])
    print()


def dls(state, depth, path, visited):

    if state == goal:
        print("Goal State Reached!")
        print_puzzle(state)
        print("Moves:", path)
        return True

    if depth == 0:
        return False

    visited.add(state)

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [
        ("Down", 1, 0),
        ("Up", -1, 0),
        ("Right", 0, 1),
        ("Left", 0, -1)
    ]

    for move, dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = list(state)

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            new_state = tuple(new_state)

            if new_state not in visited:

                print("Move:", move)
                print_puzzle(new_state)

                if dls(new_state, depth - 1,
                       path + [move], visited):
                    return True

    visited.remove(state)

    return False


def ids(start):

    for depth in range(20):

        print("Depth Limit:", depth)

        visited = set()

        if dls(start, depth, [], visited):
            return

    print("Solution not found")


print("Initial State:")
print_puzzle(start)

ids(start)
```
