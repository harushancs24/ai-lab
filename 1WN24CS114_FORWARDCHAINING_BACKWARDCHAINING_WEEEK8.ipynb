```python



facts = {
    "Barks(Tommy)",
    "FourLegs(Tommy)",
    "SharpTeeth(Tommy)"
}


rules = [
    (["Barks(Tommy)", "FourLegs(Tommy)"], "Dog(Tommy)"),
    (["Feathers(Tommy)", "LaysEggs(Tommy)"], "Bird(Tommy)"),
    (["Dog(Tommy)"], "Mammal(Tommy)"),
    (["Mammal(Tommy)", "SharpTeeth(Tommy)"], "Carnivore(Tommy)"),
    (["Bird(Tommy)"], "CanFly(Tommy)")
]


def forward_chaining(facts, rules, goal):

    while True:
        new_fact_found = False

        for conditions, conclusion in rules:


            if all(condition in facts for condition in conditions):

                if conclusion not in facts:
                    facts.add(conclusion)
                    print("Derived:", conclusion)
                    new_fact_found = True


                if goal in facts:
                    return True


        if not new_fact_found:
            break

    return goal in facts



goal = "Carnivore(Tommy)"


if forward_chaining(facts, rules, goal):
    print("\nGoal proved: Tommy is a Carnivore")
else:
    print("\nGoal cannot be proved")


```

```python




facts = {
    "Croaks(Fritz)",
    "EatsFlies(Fritz)"
}


rules = [
    (["Croaks(Fritz)", "EatsFlies(Fritz)"], "Frog(Fritz)"),
    (["Chirps(Fritz)", "Sings(Fritz)"], "Canary(Fritz)"),
    (["Frog(Fritz)"], "Green(Fritz)"),
    (["Canary(Fritz)"], "Yellow(Fritz)")
]


def backward_chaining(goal):


    if goal in facts:
        return True


    for conditions, conclusion in rules:

        if conclusion == goal:


            if all(backward_chaining(condition)
                   for condition in conditions):
                return True


    return False



goal = "Green(Fritz)"


if backward_chaining(goal):
    print("Goal proved: Fritz is Green")
else:
    print("Goal cannot be proved")
```
