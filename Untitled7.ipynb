```python
import random

board = [' '] * 9

def display():
    print()
    print(board[0], '|', board[1], '|', board[2])
    print('--+---+--')
    print(board[3], '|', board[4], '|', board[5])
    print('--+---+--')
    print(board[6], '|', board[7], '|', board[8])
    print()

def check_winner(player):
    wins = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in wins:
        if board[a] == board[b] == board[c] == player:
            return True

    return False

def computer_move():
    empty = []

    for i in range(9):
        if board[i] == ' ':
            empty.append(i)

    return random.choice(empty)

print("Tic-Tac-Toe")
print("You are X")
print("Computer is O")

for turn in range(9):

    display()

    position = int(input("Enter your position (1-9): ")) - 1

    if position < 0 or position > 8 or board[position] != ' ':
        print("Invalid position!")
        continue

    board[position] = 'X'

    if check_winner('X'):
        display()
        print("You win!")
        break

    if ' ' not in board:
        display()
        print("Game Draw!")
        break

    position = computer_move()
    board[position] = 'O'

    print("Computer chose:", position + 1)

    if check_winner('O'):
        display()
        print("Computer wins!")
        break

else:
    display()
    print("Game Draw!")
```

```python
room = {
    'A': 'Dirty',
    'B': 'Dirty'
}

location = 'A'

while True:
    print("Location:", location)
    print("Room Status:", room[location])

    if room[location] == 'Dirty':
        print("Action: Clean")
        room[location] = 'Clean'

    else:
        if location == 'A':
            location = 'B'
            print("Action: Move Right")
        else:
            location = 'A'
            print("Action: Move Left")

    print()

    if room['A'] == 'Clean' and room['B'] == 'Clean':
        print("Both rooms are clean.")
        break
```
